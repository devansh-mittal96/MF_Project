# Day 3 — Exploratory Data Analysis\n\nNAV trends, investor demographics, correlation.


```python
from pathlib import Path
import sys

def find_root():
    cwd = Path.cwd()
    if (cwd / "data" / "raw").exists():
        return cwd
    if (cwd.parent / "data" / "raw").exists():
        return cwd.parent
    raise FileNotFoundError("Open notebook from project root or notebooks/ folder.")

ROOT = find_root()
sys.path.insert(0, str(ROOT / "scripts"))
print("Project root:", ROOT)

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine

engine = create_engine(f"sqlite:///{ROOT / 'data' / 'db' / 'bluestock_mf.db'}")
nav = pd.read_sql("SELECT * FROM fact_nav", engine)
nav['date'] = pd.to_datetime(nav['date'])
code = 'amfi_code' if 'amfi_code' in nav.columns else 'scheme_code'
plt.figure(figsize=(12,5))
for c in nav[code].unique():
    s = nav[nav[code]==c]
    plt.plot(s['date'], s['nav'], label=str(c))
plt.legend(); plt.title('NAV Trend'); plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'charts' / 'eda_nav_trend.png', dpi=150)
plt.show()

```
